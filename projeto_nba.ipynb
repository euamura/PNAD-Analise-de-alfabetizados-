## Consulta e Cruzamentos em SQL:

```python
# criar função de cruzamento de dados
def cruzamento_dados(query, new_table):
    df_resultado = pd.read_sql_query(query, conn_origem)
    df_resultado.to_sql(new_table, conn_destino, if_exists="replace", index=False)
    print(f"Tabela '{new_table}' criada no banco de destino com dados cruzados.")
```

```python
# criar tabela e cruzar dados para banco de destino

## players
query1 = '''
  SELECT
    p.id AS player_id,
    p.full_name AS player_name,
    cp.team_id,
    cp.position,
    cp.height,
    cp.birthdate
  FROM player p
  JOIN common_player_info cp ON p.id = cp.person_id
'''

cruzamento_dados(query1, 'players')

## teams
query2 = '''
  SELECT
    t.id AS team_id,
    t.full_name AS team_name,
    t.city,
    td.arena
  FROM team t
  JOIN team_details td ON t.id = td.team_id
'''

cruzamento_dados(query2, 'teams')

## games
query3 = '''
  SELECT
    g.game_id,
    g.team_id_home AS team_id,
    g.game_date,
    g.pts_home AS score,
    g.wl_home AS win_lose,
    g.ast_home AS assists,
    g.reb_home AS rebounds
  FROM game g
'''

cruzamento_dados(query3, 'games')

```

```python
# exibir dados tabela no bd de destino
cursor_destino.execute("PRAGMA table_info('players')")
print("tabela players:", cursor_destino.fetchall())

cursor_destino.execute("PRAGMA table_info('teams')")
print("tabela teams:", cursor_destino.fetchall())

cursor_destino.execute("PRAGMA table_info('games')")
print("tabela games:", cursor_destino.fetchall())
```

```python
# criar tabela com união de dados do bd de destino

cursor_destino.execute( '''
CREATE TABLE IF NOT EXISTS general_table AS
  SELECT
    g.*,
    t.team_name,
    t.city,
    t.arena,
    p.player_id,
    P.player_name,
    P.position,
    P.height,
    P.birthdate
  FROM games g
  JOIN teams t ON g.team_id = t.team_id
  JOIN players p ON g.team_id = p.team_id
  WHERE game_date >= DATE('now', '-10 years');

''')
conn_destino.commit()
```

```python
cursor_destino.execute("PRAGMA table_info('general_table')")
print("tabela general_table:", cursor_destino.fetchall())
```
